```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
# load the data 
df = pd.read_csv('../Data/telco_comm_churn.csv')
df.head()
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
df.describe()
```

## handling missing values

```python
df.isnull().sum()
```

```python
df.fillna({'TotalCharges': df['TotalCharges'].median()},inplace=True)
```

```python
df.isnull().sum()
```

## Feature engeenering
#### creating new features

```python
df['tenure'].min(), df['tenure'].max()
# df['tenure']
```

```python
df['tenure_group'] = pd.cut(df['tenure'],
       bins=[-1,12,24,48,73],
       labels=[0,1,2,3],
       include_lowest= True)
```

```python
df.head()
```

```python
df[['tenure','tenure_group']]
```

```python
df['tenure_group'].info()
```

```python
df['tenure_group'] = df['tenure_group'].astype(float).astype(int)
df['tenure_group'].info()
```

```python
df['avg_monthly_per_tenure'] = df['TotalCharges']/ (df['tenure']+ 1)
```

```python
df.columns
```

## Encoding

```python
df = df.drop('customerID', axis=1)
```

```python
df.head(2)
```

```python
binary_cols = ['Partner','Dependents','PhoneService','PaperlessBilling']
for col in binary_cols:
    df[col] = df[col].map({'Yes':1 , 'No':0})
df.head()
```

```python
df['Churn'] = df['Churn'].map({'Yes':1 , 'No':0})
```

```python
df['gender'] = df['gender'].map({'Male':1 , 'Female':0})
```

```python
df.info()
```

```python
catagorical_cols = df.select_dtypes(include=['object']).columns.tolist()
```

```python
# df[catagorical_cols].count()
for col in catagorical_cols:
    le =  LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
```
